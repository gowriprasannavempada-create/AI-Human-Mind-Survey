# AI Technologies and Human Mind – Survey Analysis

## Introduction

Artificial Intelligence (AI) has become an important part of education, work, research, and everyday activities. Many people use AI tools for learning, coding, searching for information, completing tasks, and saving time.

This project analyzes survey responses collected through a Google Form to understand how people use AI and how they feel AI affects different aspects of the human mind. The analysis focuses on AI usage, reasons for using AI, productivity, learning, creativity, independent thinking, dependency on technology, trust in AI-generated information, verification habits, benefits, concerns, and expected future impact.

The survey contains responses from 50 participants. Python, Pandas, Matplotlib, and Seaborn are used to clean, analyze, and visualize the survey data.

The main objective of this project is to understand the different ways people interact with AI and identify both the positive and negative aspects of AI based on the survey responses.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("final_AI_survey_analysis.csv")

print("Rows:", data.shape[0])
print("Columns:", data.shape[1])
```

Analyze AI Usage

```python
tools = data[
    "6. Which AI tools have you used?"
].dropna().str.split(",")

tools = tools.explode().str.strip()

tool_counts = tools.value_counts()

print(tool_counts)
```

```python
plt.figure(figsize=(10,6))

tool_counts.plot(kind="bar")

plt.title("AI Tools Used by Respondents")
plt.xlabel("AI Tool")
plt.ylabel("Number of Users")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

Analyze Why People Use AI

```python
reasons = data[
    "9. Why do you use AI?"
].dropna().str.split(",")

reasons = reasons.explode().str.strip()

reason_counts = reasons.value_counts()

print(reason_counts)
```

```python
plt.figure(figsize=(10,6))

reason_counts.plot(kind="bar")

plt.title("Reasons for Using AI")
plt.xlabel("Reason")
plt.ylabel("Number of Responses")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

Human Mind Analysis

```python
impact_columns = [
    "10. AI helps me learn faster.",
    "11. AI improves my productivity.",
    "13. AI has made me more dependent on technology.",
    "14. AI has improved my creativity.",
    "15. AI has reduced my independent thinking.",
    "20. AI helps me complete tasks more quickly."
]

impact_average = data[impact_columns].mean()

print(impact_average)
```

```python
plt.figure(figsize=(12,6))

impact_average.plot(kind="bar")

plt.title("Impact of AI on Human Activities")
plt.ylabel("Average Rating")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

Trust vs verification

```python
trust_mapping = {
    "Never": 1,
    "Rarely": 2,
    "Sometimes": 3,
    "Mostly": 4,
    "Always": 5
}

data["Trust_Score"] = data[
    "17. How much do you trust AI-generated information?"
].map(trust_mapping)
```

```python
print(data[[
    "17. How much do you trust AI-generated information?",
    "Trust_Score"
]].head(10))
```

```python
trust_verification = data.groupby(
    "16. Do you verify AI-generated information before using it?"
)["Trust_Score"].mean()

print(trust_verification)
```

```python
plt.figure(figsize=(8,5))

trust_verification.plot(kind="bar")

plt.title("Verification vs Trust in AI")
plt.xlabel("Verification")
plt.ylabel("Average Trust Score")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

Key Findings

* The survey collected responses from 50 participants about their use and perception of Artificial Intelligence.
* ChatGPT was the most commonly used AI tool among the respondents.
* Studying was the main reason for using AI, followed by coding and research.
* AI received the highest average rating for helping users complete tasks more quickly, with an average rating of 4.28.
* AI also showed high ratings for helping users learn faster (4.10) and improving productivity (4.04).
* The average rating for dependency on technology was 3.70, showing that respondents also noticed some level of dependence on AI and technology.
* Independent thinking had an average rating of 3.44, while creativity had an average rating of 3.42.
* Most respondents reported that they trust AI-generated information either "Mostly" or "Sometimes."
* Most respondents verify AI-generated information at least sometimes before using it.
* Job loss and overdependence on technology were among the most frequently reported concerns about AI.
* Saving time was the most commonly selected benefit of AI.
* Software development and education were the areas where respondents most frequently expected AI to have a future impact.


# Conclusion

The survey shows that AI has become an important part of the respondents' daily activities, especially for studying, coding, research, and completing tasks. The responses indicate that AI is mainly viewed as a useful tool for saving time, learning faster, and improving productivity.

At the same time, the survey shows that respondents are aware of possible negative effects such as dependence on technology, reduced independent thinking, job loss, privacy issues, and security risks. Trust in AI-generated information is present, but many respondents still verify information before using it.

Overall, the survey indicates that AI can provide significant benefits when used as a supporting tool. The responses also highlight the importance of verifying AI-generated information and maintaining independent thinking while using AI.
